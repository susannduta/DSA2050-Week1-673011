## Data Visualization

```python
# Revenue by Region
region_revenue = df.groupby("Region")["Revenue_KES"].sum()

region_revenue.plot(kind="bar", figsize=(8, 5))

plt.title("Total Revenue by Region")
plt.xlabel("Region")
plt.ylabel("Total Revenue (KES)")
plt.xticks(rotation=0)
plt.show()
```

Top-Performing Region: Kisumu generated the highest total revenue, at approximately KES 880,000.

Second-Highest Region: Mombasa followed with total revenue of approximately KES 800,000.

Other Regions: Nakuru and Nairobi recorded slightly lower revenues of approximately KES 780,000 and KES 760,000, respectively, indicating relatively close revenue levels among the four regions.

```python
# Revenue by product category
category_revenue = df.groupby("Category")["Revenue_KES"].sum()

category_revenue.plot(kind="bar", figsize=(8, 5))

plt.title("Total Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Revenue (KES)")
plt.xticks(rotation=0)
plt.show()
```

Top-Performing Categories: Electronics, Personal Care, and Home & Office are the leading revenue-generating categories, each contributing approximately KES 920,000–930,000 in total revenue. Their relatively similar performance indicates a balanced contribution to overall sales.

Lowest-Performing Category: Accessories generated the lowest revenue, at approximately KES 420,000, placing it substantially below the other three categories in terms of total sales contribution.
